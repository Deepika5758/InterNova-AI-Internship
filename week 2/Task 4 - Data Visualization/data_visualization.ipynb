```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned student dataset
df = pd.read_csv("../Task 3 - Data Cleaning/student_performance_cleaned.csv")

print("Cleaned Student Dataset:")
print(df)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Student_Name"], df["Python_Marks"])

plt.title("Student Python Marks")
plt.xlabel("Student Name")
plt.ylabel("Python Marks")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Student_Name"], df["Maths_Marks"], marker="o")

plt.title("Student Mathematics Performance")
plt.xlabel("Student Name")
plt.ylabel("Mathematics Marks")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Python_Marks"], bins=5, edgecolor="black")

plt.title("Distribution of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Python_Marks"], df["Maths_Marks"])

plt.title("Python Marks vs Mathematics Marks")
plt.xlabel("Python Marks")
plt.ylabel("Mathematics Marks")

plt.tight_layout()
plt.show()
```
