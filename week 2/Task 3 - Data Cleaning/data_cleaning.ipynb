```python
import pandas as pd

# Load uncleaned dataset
df = pd.read_csv("student_performance_uncleaned.csv")

print("Uncleaned Dataset:")
print(df)
```

```python
# Identify missing values
print("Missing Values:")
print(df.isnull())

# Count missing values in each column
print("\nMissing Values Count:")
print(df.isnull().sum())
```

```python
# Identify duplicate records
print("Number of Duplicate Records:")
print(df.duplicated().sum())

print("\nDuplicate Records:")
print(df[df.duplicated()])
```

```python
# Fill missing values with the mean of each column
df["Python_Marks"] = df["Python_Marks"].fillna(
    df["Python_Marks"].mean()
)

df["Science_Marks"] = df["Science_Marks"].fillna(
    df["Science_Marks"].mean()
)

df["Attendance"] = df["Attendance"].fillna(
    df["Attendance"].mean()
)

print("Missing values after filling:")
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate records after removal:")
print(df.duplicated().sum())

print("\nCleaned Dataset:")
print(df)
```

```python
df["Student_ID"] = df["Student_ID"].astype(int)

print("Updated Data Types:")
print(df.dtypes)
```

```python
df.to_csv("student_performance_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
