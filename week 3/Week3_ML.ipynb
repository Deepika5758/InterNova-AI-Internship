```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Study_Hours": [2,3,5,7,8,1,4,6,9,3,
                    5,7,2,8,6,4,9,1,5,7,
                    3,6,8,2,4,9,5,7,1,6],

    "Attendance": [45,55,75,90,95,30,65,85,98,50,
                   70,88,40,92,80,60,96,35,72,89,
                   52,82,94,42,62,99,74,87,32,84],

    "Previous_Marks": [35,42,65,80,88,25,55,72,95,40,
                       68,82,30,90,75,50,96,28,66,85,
                       45,78,92,32,58,97,70,83,20,79],

    "Assignment_Score": [30,45,60,85,92,20,50,78,96,42,
                         65,88,35,90,80,55,98,25,70,86,
                         48,75,94,38,62,99,72,89,22,81]
}

df = pd.DataFrame(data)

# Generate PASS/FAIL labels for this demonstration dataset
score = (
    0.15 * (df["Study_Hours"] / 10 * 100)
    + 0.25 * df["Attendance"]
    + 0.30 * df["Previous_Marks"]
    + 0.30 * df["Assignment_Score"]
)

df["Result"] = np.where(score >= 50, "PASS", "FAIL")

# Save the dataset as a CSV file
df.to_csv("Dataset.csv", index=False)

print("Dataset created successfully!")
```

# Task 4: Build Your First Machine Learning Model

In this task, we build a Decision Tree Classifier using Scikit-learn. The model learns from student data and predicts whether a student is likely to PASS or FAIL.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)
```

```python
model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print("Actual Results:")
print(y_test.values)

print("\nPredicted Results:")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy * 100, "%")
```

```python
results = pd.DataFrame({
    "Actual Result": y_test.values,
    "Predicted Result": y_pred
})

display(results)
```
