```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the cleaned student dataset
df = pd.read_csv("../Task 3 - Data Cleaning/student_performance_cleaned.csv")

print("STUDENT PERFORMANCE DATA ANALYSIS")
print("----------------------------------")

print("\nStudent Dataset:")
print(df)
```

```python
print("Dataset Information:")
print(df.info())

print("\nNumber of Rows and Columns:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# Calculate average marks for each student
df["Average_Marks"] = df[
    ["Python_Marks", "Maths_Marks", "Science_Marks"]
].mean(axis=1)

# Find highest and lowest average marks
highest_marks = df["Average_Marks"].max()
lowest_marks = df["Average_Marks"].min()

# Count passed and failed students
passed = (
    (df["Python_Marks"] >= 40) &
    (df["Maths_Marks"] >= 40) &
    (df["Science_Marks"] >= 40)
).sum()

failed = len(df) - passed

print("STUDENT PERFORMANCE RESULTS")
print("----------------------------")
print("Overall Average Marks:", round(df["Average_Marks"].mean(), 2))
print("Highest Average Marks:", highest_marks)
print("Lowest Average Marks:", lowest_marks)
print("Number of Students Passed:", passed)
print("Number of Students Failed:", failed)

print("\nStudent-wise Average Marks:")
print(df[["Student_Name", "Average_Marks"]])
```

```python
plt.figure(figsize=(9, 5))

plt.bar(df["Student_Name"], df["Average_Marks"])

plt.title("Average Marks of Students")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 6))

plt.pie(
    [passed, failed],
    labels=["Passed", "Failed"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Pass and Fail Percentage")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Python_Marks"], df["Science_Marks"])

plt.title("Python Marks vs Science Marks")
plt.xlabel("Python Marks")
plt.ylabel("Science Marks")

plt.tight_layout()
plt.show()
```
